Assignment 35: Fibonacci Function

Write a function named generate_fibonacci().

Requirements

Accept the number of terms.

Return the complete Fibonacci series as a list.

Handle zero and negative term counts.

Do not use recursion in this assignment.

```python
def generate_fibonacci(terms):
  if terms <= 0:
    return []
  elif terms == 1:
    return [0]
  fibonacci = [0, 1]
  first = 0
  second = 1
  for i in range(terms - 2):
    next_number = first + second
    fibonacci.append(next_number)
    first = second
    second = next_number
  return fibonacci
n = int(input("enter number of terms: "))
res = generate_fibonacci(n)
res
```

Assignment 92: Closure-Based Bank Balance

Store the balance in the outer function.

Define inner functions for:

Deposit

Withdrawal

Balance enquiry

Use nonlocal to update the balance.

Reject withdrawals greater than the available balance.

Return all three inner functions.

```python
def create_bank_account(balance):
  def deposit(amount):
    nonlocal balance
    balance += amount
    print("Balance after deposit:", balance)
  def withdrawl(amount):
    nonlocal balance
    if amount <= balance:
      balance -= amount
      print("Balance after withdrawl:", balance)
    else:
      print("Insufficient balance")
  def balance_enquiry():
    print("Current balance:", balance)
  return deposit, withdrawl, balance_enquiry
deposit, withdrawl, balance_enquiry =  create_bank_account(1000)
balance_enquiry()
deposit(500)
withdrawl(300)
withdrawl(2000)
balance_enquiry()

```

Starting function...

Result: 30

Function completed.

```python
# part5
def calculate_revised_salaries(emp_list):
  def revise_salary(emp):
    update_emp = emp.copy()
    score = emp["performance_score"]
    if score > 90:
      update_emp["salary"] = emp["salary"] * 1.15
    elif score > 80:
      update_emp["salary"] = emp["salary"]  * 1.10
    elif score > 70:
      update_emp["salary"] = emp["salary"] * 1.05
    else:
      update_emp["salary"] = emp["salary"]
    return update_emp
  revised_employees = list(map(revise_salary, emp_list))
  for emp in revised_employees:
    print("\nName: ", emp["name"])
    print("Salary:", emp_list[revised_employees.index(emp)]["salary"])
    print("Updated Salary:", emp["salary"])
  return revised_employees
calculate_revised_salaries(employees)
```

```python
from functools import reduce
def calculate_total_payroll(emp_list, revised_list):
  current_total_payroll = reduce(lambda total, emp: total + emp["salary"], emp_list, 0)
  revised_total_payroll = reduce(lambda total, emp: total + emp["salary"], revised_list, 0)
  additional_expense = revised_total_payroll - current_total_payroll
  print("Current salary:", current_total_payroll)
  print("Revised salary:", revised_total_payroll)
  print("Additional Expenses:", additional_expense)
revised_employees= calculate_revised_salaries(employees)
calculate_total_payroll(employees, revised_employees)
```

```python
def generate_department_report(emp_list, department):
    dept_emp = list(filter(lambda emp: emp["department"].lower() == department.lower(), emp_list))
    if not dept_emp:
        return {"Department not found"}
    def emp_count():
        return len(dept_emp)
    def total_salary():
        return sum(emp["salary"] for emp in dept_emp)
    def avg_salary():
        return total_salary() / emp_count()
    def avg_performance():
        total_score = sum(emp["performance_score"] for emp in dept_emp)
        return total_score / emp_count()
    report = {
        "Department": department,
        "Number of Employees": emp_count(),
        "Total Salary": total_salary(),
        "Average Salary": avg_salary(),
        "Average Performance Score": avg_performance()
    }
    return report
department = input("Enter Department: ")
report = generate_department_report(employees, department)
for key, value in report.items():
    print(key, ":", value)
```

```python
company_name = "Edukron"         # global
def generate_department_report(emp_list, department):
    dept_name = department #enclosing
    dept_emp = list(filter(lambda emp: emp["department"].lower() == dept_name.lower(), emp_list))
    if not dept_emp:
        return { "error":"Department not found"}
    def create_report():
      report_title = "Department Report"
      return {
            "Company": company_name,          # Global variable
            "Title": report_title,            # Local variable
            "Department": dept_name,    # Enclosing variable
            "Employees": len(dept_emp),
            "Total Salary": sum(emp["salary"] for emp in dept_emp),
            "Avg salary": round(sum(emp["salary"] for emp in dept_emp) / len(dept_emp), 2 ),
            "Avg Performance": round(sum(emp["performance_score"] for emp in dept_emp) / len(dept_emp), 2)
       }
    return create_report()
department = input("Enter Department: ")
report = generate_department_report(employees, department)
for key, value in report.items():
    print(key, ":", value)
```
