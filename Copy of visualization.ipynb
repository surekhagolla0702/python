```python
import matplotlib.pyplot as plt
x = [1, 2, 3, 4, 5]
y = [20, 40, 60, 80, 100]
plt.plot(x, y,color = 'green', marker='o', markerfacecolor = 'r', markeredgecolor = 'black', linewidth = 1)
plt.title("Line plot")
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True)
plt.show
```

Insights from the graph:
The graph shows a positive linear relationship between x and y.
As x increases, y also increases consistently.
For every 1-unit increase in x, y increases by 20 units.
There are no sudden increases or decreases, so the trend is consistent.
The relationship can be represented approximately as y = 20x.
The highest value is y = 100 at x = 5, while the lowest is y = 20 at x = 1.

Overall insight: The graph represents a strong, perfectly linear increasing trend between x and y.

```python
import seaborn as sns
import matplotlib.pyplot as plt
flights = sns.load_dataset('flights')
sns.lineplot(data=flights, x='year', y='passengers')
plt.title('Passengers over Time')
plt.show()
```

overall upward trend: The number of passengers increases steadily from around 125 in 1949 to approximately 475 in 1960.

Highest passengers: The peak is in 1960, with about 475 passengers.

Lowest passengers: The lowest value is in 1949, at around 125 passengers.

Strong growth: Passenger numbers increased significantly, especially between 1954 and 1957.

Small slowdown: There is a slight slowdown between 1957 and 1958, where growth is less steep.

Growth resumes: From 1958 to 1960, passengers rise again considerably.

Confidence/variation: The shaded area around the line shows some uncertainty or variation in the passenger values, which becomes wider toward the later years.

```python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset('tips')

sns.boxplot(data=tips, x='day', y='total_bill', palette='Set2')
plt.title('Bill Amount by Day')
plt.show()
```

Insights from the Box Plot 📊 — Bill Amount by Day

Sunday has the highest median bill — around 20, meaning typical bills are highest on Sunday.

Saturday has the widest spread of bill amounts, indicating greater variation in spending.

Thursday has several outliers with bills above 30, including values around 41–43.

Saturday also has high outliers, with bills reaching approximately 51, the highest bill shown.

Sunday has a high maximum of around 48, but fewer outliers compared with Thursday.

Friday has the lowest overall range, with bills generally concentrated between approximately 12 and 22.

Weekend bills tend to be higher than weekday bills, particularly on Saturday and Sunday.
