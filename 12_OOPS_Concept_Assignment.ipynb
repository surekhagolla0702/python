Q1. Create a account class for banking case 1; use sample reference ACC-1001 and explain the result.

Hints

List the objects, their responsibilities, and their invariants before coding.

Choose composition for has-a relationships; use inheritance only for genuine is-a substitution.

Keep balances private and expose controlled deposit/withdraw/transfer behavior.

Edge cases to test: empty input, invalid types, boundary values, and a realistic failure path.

```python
class Account:
    def __init__(self, ref, balance):
        self.ref = ref
        self.__balance = balance
    def deposit(self, amount):
        if amount <= 0:
            raise ValueError("Invalid amount")
        self.__balance += amount
    def withdraw(self, amount):
        if amount > self.__balance:
            raise ValueError("Insufficient balance")
        self.__balance -= amount
    def balance(self):
        return self.__balance
account = Account("ACC-1001", 5000)
try:
    amount = input("Enter amount: ")
    if amount == "":
        raise ValueError("Empty input")
    amount = float(amount)
    account.deposit(amount)
    print("Account:", account.ref)
    print("Balance:", account.balance())
except ValueError as e:
    print("Error:", e)
```
